# Add Maven Wrapper

This notebook documents the steps to add a Maven wrapper to the `college-dashboard` Java project. It includes wrapper files, configuration, and a verification build step.


## 1. Install Maven Wrapper

Use the Maven wrapper plugin to generate wrapper files in the project directory. This step is typically done with a globally installed Maven, but here we simulate the file creation manually if needed.


```python
# Normally this command is run with Maven installed:
# mvn -N io.takari:maven:wrapper

print('Maven wrapper files will be created manually in the next steps.')
```

## 2. Add Wrapper Scripts

Create the `mvnw`, `mvnw.cmd`, and `.mvn/wrapper` files to enable project builds without a global Maven install.


```python
from pathlib import Path

base = Path('d:/college-dashboard')
(base / 'mvnw').write_text("""#!/usr/bin/env sh
MAVEN_PROJECTBASEDIR=\"$(cd \"$(dirname \"$0\")\" && pwd -P)\"
exec \"${MAVEN_PROJECTBASEDIR}/.mvn/wrapper/maven-wrapper.jar\" \"$@\"
""")
(base / 'mvnw.cmd').write_text("""@echo off\r\nsetlocal enableextensions enabledelayedexpansion\r\nset \"MAVEN_PROJECTBASEDIR=%~dp0\"\r\n\"%MAVEN_PROJECTBASEDIR%\\.mvn\\wrapper\\maven-wrapper.jar\" %*\r\n""")
(base / '.mvn' / 'wrapper').mkdir(parents=True, exist_ok=True)
print('Wrapper scripts created.')
```

## 3. Configure Wrapper Properties

Edit `.mvn/wrapper/maven-wrapper.properties` to specify the Maven distribution version and download URL.


```python
from pathlib import Path

props = Path('d:/college-dashboard/.mvn/wrapper/maven-wrapper.properties')
props.write_text("""distributionUrl=https://repo1.maven.org/maven2/org/apache/maven/apache-maven/3.9.9/apache-maven-3.9.9-bin.zip
wrapperUrl=https://repo1.maven.org/maven2/org/apache/maven/wrapper/maven-wrapper/3.3.4/maven-wrapper-3.3.4.jar
""")
print('Wrapper properties configured.')
```

## 4. Run Maven Wrapper Build

Use `./mvnw` or `mvnw.cmd` to build the project and verify that the wrapper works correctly. If the Maven wrapper jar is missing, the wrapper will download it automatically from Maven Central.


```python
import subprocess
from pathlib import Path

project_dir = Path('d:/college-dashboard')
result = subprocess.run([str(project_dir / 'mvnw.cmd'), 'clean', 'package'], cwd=project_dir, capture_output=True, text=True)
print('STDOUT:\n', result.stdout)
print('STDERR:\n', result.stderr)
print('Return code:', result.returncode)
```
